```python
import numpy as np
import pandas as pd
import os
```

```python
chroms=list(range(1, 23))
chroms=[str(x) for x in chroms]
```

```python
chroms
```

```python
# check overlap between low coverage and high coverage biallelic snp sites in 1KG
num_total_pos_lc = []
num_total_pos_hc = []
lc_num_in_hc = []

for chrom in chroms:
    hc = pd.read_csv(f'/wynton/home/capra/gramey02/dnd_project/data/biallelic_afs_high_coverage/TGP_chr{chrom}_afs.txt',
                      sep='\t', names=['chrom','pos','ref','alt','ac','an','af','af1','af2','af3','af4','af5','af6','af7'])
    lc = pd.read_csv(f'/wynton/home/capra/gramey02/dnd_project/data/biallelic_afs/TGP_chr{chrom}_afs.txt',
                      sep=' ', names=['chrom','pos','ref','alt','af','ac','an','af1','af2','af3','af4','af5','af6','af7'])
    print(f'loaded data frames for chromosome {chrom}')
    hc_pos = hc.pos.unique()
    lc_pos = lc.pos.unique()

    # check overlap in positions
    num_total_pos_lc.append(len(lc_pos))
    num_total_pos_hc.append(len(hc_pos))
    lc_num_in_hc.append(pd.Series(lc_pos).isin(hc_pos).sum())
    print(f'calculated overlap for chromsome {chrom}')
```

```python
# calculate fraction of low coverage vars that show up in high coverage files
frac_lc_in_hc = [overlap / total for overlap, total in zip(lc_num_in_hc, num_total_pos_lc)]
```
