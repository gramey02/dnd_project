## gRNA mismatch analysis
This will tell us how many gRNAs per gene have a gRNA that might tolerate a single base pair mismatch and therefore not be allele-specific in the way we wish

```python
import numpy as np
import pandas as pd
import pickle
import os
```

```python
results_dir = '/wynton/home/capra/gramey02/dnd_project/results'
run_name = 'RUN_MULTIALLELIC'
strats=['indels','CRISPRoff','base_edits','excision']
excavate_dir='excavate/excavate_outputs'
```

```python
results={}
proportion_results={}
for strat in strats:
    strat_result={}
    strat_prop_results={}
    if strat!='base_edits':
        cur_fp = os.path.join(results_dir, run_name, strat, excavate_dir)
        cur_outputs = os.listdir(cur_fp)
        for file in cur_outputs:
            cur_gene=file.split('_')[0]
            # load the gene's gRNA outputs
            if os.path.exists(os.path.join(cur_fp, file, 'all_guides.csv')):
                grnas = pd.read_csv(os.path.join(cur_fp, file, 'all_guides.csv'))
            else:
                continue
            # not how many gRNAs would tolerate a mismatch
            total_grnas=len(grnas)
            mismatch_grnas = grnas[(grnas['strand']=='+') & 
                (grnas['variation'].isin(['A>G','C>T'])) & 
                (grnas['present allele'].str.contains('(alt)', regex=False))]
            total_mismatch_grnas = len(mismatch_grnas)

            # store the results
            strat_result[cur_gene]={'total_grnas':total_grnas,'total_mismatch_grnas':total_mismatch_grnas}
            if total_grnas>0:
                strat_prop_results[cur_gene]=total_mismatch_grnas/total_grnas
            else:
                strat_prop_results[cur_gene]=np.nan
    else:
        cur_fp_acc=os.path.join(results_dir, run_name, 'acceptor_base_edits', excavate_dir)
        cur_fp_donor=os.path.join(results_dir, run_name, 'donor_base_edits', excavate_dir)

        cur_acc_outputs=os.listdir(cur_fp_acc)
        cur_donor_outputs=os.listdir(cur_fp_donor)

        acc_genes = {x.split('_')[0] for x in cur_acc_outputs}
        donor_genes = {x.split('_')[0] for x in cur_donor_outputs}
        all_base_edit_genes = acc_genes | donor_genes

        for gene in all_base_edit_genes:
            acc_grnas = None
            donor_grnas = None
            if os.path.exists(os.path.join(cur_fp_acc, gene + '_output', 'all_guides.csv')):
                acc_grnas = pd.read_csv(os.path.join(cur_fp_acc, gene + '_output', 'all_guides.csv'))
            if os.path.exists(os.path.join(cur_fp_donor, gene + '_output', 'all_guides.csv')):
                donor_grnas = pd.read_csv(os.path.join(cur_fp_donor, gene + '_output', 'all_guides.csv'))

            grna_frames = [df for df in (acc_grnas, donor_grnas) if df is not None]
            if not grna_frames:
                continue  # skip gene if neither file exists
            base_edit_grnas = pd.concat(grna_frames).drop_duplicates()
            
            total_grnas=len(base_edit_grnas)
            mismatch_grnas = base_edit_grnas[(base_edit_grnas['strand']=='+') & 
                (base_edit_grnas['variation'].isin(['A>G','C>T'])) & 
                (base_edit_grnas['present allele'].str.contains('(alt)', regex=False))]
            total_mismatch_grnas = len(mismatch_grnas)
            strat_result[gene]={'total_grnas':total_grnas,'total_mismatch_grnas':total_mismatch_grnas}
            if total_grnas>0:
                strat_prop_results[gene]=total_mismatch_grnas/total_grnas
            else:
                strat_prop_results[gene]=np.nan
    
    results[strat]=strat_result
    proportion_results[strat]=strat_prop_results
```

```python
proportion_results
```

```python
results
```

```python
# then let's get a readout of per-gene/strategy combination, how often this happens:
per_strat_prop = {}
for strat, strat_result in proportion_results.items():
    per_strat_prop[strat] = np.nanmean(list(strat_result.values()))
```

```python
per_strat_prop
```

```python
np.mean(list(per_strat_prop.values()))
```

## This means that generally about a tenth of the gRNAs have this mismatch tolerance

## Let's also check whether, if this occurs, if the same snp has a non-mismatch tolerating gRNA as an option
e.g., like (some sort of more reliable backup option)

```python
snp_results={}
for strat in strats:
    strat_result={}
    if strat!='base_edits':
        cur_fp = os.path.join(results_dir, run_name, strat, excavate_dir)
        cur_outputs = os.listdir(cur_fp)
        for file in cur_outputs:
            cur_gene=file.split('_')[0]
            # load the gene's gRNA outputs
            if os.path.exists(os.path.join(cur_fp, file, 'all_guides.csv')):
                grnas = pd.read_csv(os.path.join(cur_fp, file, 'all_guides.csv'))
            else:
                continue
            # note how many gRNAs would tolerate a mismatch
            total_grnas=len(grnas)
            mismatch_grnas = grnas[(grnas['strand']=='+') & 
                (grnas['variation'].isin(['A>G','C>T'])) & 
                (grnas['present allele'].str.contains('(alt)', regex=False))]
            total_mismatch_grnas = len(mismatch_grnas)

            # check which snp position this pertains to
            mismatch_snps = set(mismatch_grnas['SNP position'])
            # check if there are non-mismatch grnas for these same snps
            total_mismatch_snps=len(mismatch_snps)
            counter=0
            if total_mismatch_snps>0:
                for cur_snp in mismatch_snps:
                    non_mismatch_grnas = grnas[~((grnas['strand']=='+') & 
                        (grnas['variation'].isin(['A>G','C>T'])) & 
                        (grnas['present allele'].str.contains('(alt)', regex=False))) & (grnas['SNP position']==cur_snp)]
                    if len(non_mismatch_grnas)>0:
                        counter+=1
                strat_result[cur_gene]=counter/total_mismatch_snps


    else:
        cur_fp_acc=os.path.join(results_dir, run_name, 'acceptor_base_edits', excavate_dir)
        cur_fp_donor=os.path.join(results_dir, run_name, 'donor_base_edits', excavate_dir)

        cur_acc_outputs=os.listdir(cur_fp_acc)
        cur_donor_outputs=os.listdir(cur_fp_donor)

        acc_genes = {x.split('_')[0] for x in cur_acc_outputs}
        donor_genes = {x.split('_')[0] for x in cur_donor_outputs}
        all_base_edit_genes = acc_genes | donor_genes

        for gene in all_base_edit_genes:
            acc_grnas = None
            donor_grnas = None
            if os.path.exists(os.path.join(cur_fp_acc, gene + '_output', 'all_guides.csv')):
                acc_grnas = pd.read_csv(os.path.join(cur_fp_acc, gene + '_output', 'all_guides.csv'))
            if os.path.exists(os.path.join(cur_fp_donor, gene + '_output', 'all_guides.csv')):
                donor_grnas = pd.read_csv(os.path.join(cur_fp_donor, gene + '_output', 'all_guides.csv'))

            grna_frames = [df for df in (acc_grnas, donor_grnas) if df is not None]
            if not grna_frames:
                continue  # skip gene if neither file exists
            base_edit_grnas = pd.concat(grna_frames).drop_duplicates()
            
            total_grnas=len(base_edit_grnas)
            mismatch_grnas = base_edit_grnas[(base_edit_grnas['strand']=='+') & 
                (base_edit_grnas['variation'].isin(['A>G','C>T'])) & 
                (base_edit_grnas['present allele'].str.contains('(alt)', regex=False))]
            total_mismatch_grnas = len(mismatch_grnas)

            # check which snp position this pertains to
            mismatch_snps = set(mismatch_grnas['SNP position'])
            # check if there are non-mismatch grnas for these same snps
            total_mismatch_snps=len(mismatch_snps)
            counter=0
            if total_mismatch_snps>0:
                for cur_snp in mismatch_snps:
                    non_mismatch_grnas = base_edit_grnas[~((grnas['strand']=='+') & 
                        (base_edit_grnas['variation'].isin(['A>G','C>T'])) & 
                        (base_edit_grnas['present allele'].str.contains('(alt)', regex=False))) & (base_edit_grnas['SNP position']==cur_snp)]
                    if len(non_mismatch_grnas)>0:
                        counter+=1
                strat_result[gene]=counter/total_mismatch_snps
    
    snp_results[strat]=strat_result
```

```python
snp_results
```

```python
# then let's get a readout of on average across genes, for each snp that has a gRNA that tolerates a mismatch,
# does it have another grna that does not?:
per_strat_snp_prop = {}
for strat, strat_result in snp_results.items():
    per_strat_snp_prop[strat] = np.nanmean(list(strat_result.values()))
```

```python
per_strat_snp_prop
```

```python
np.mean(list(per_strat_snp_prop.values()))
```
