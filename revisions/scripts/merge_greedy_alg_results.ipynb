```python
import numpy as np
import os
import pickle
import pandas as pd
from tqdm.auto import tqdm
```

```python
# set some global variables
results_dir = '/wynton/home/capra/gramey02/dnd_project/results'
run_name = 'RUN_MULTIALLELIC'
strats=['indels','CRISPRoff','base_edits','excision']
excavate_dir='excavate/excavate_outputs'
```

```python
# load greedy algorithm results
non_excision_path = '/wynton/home/capra/gramey02/dnd_project/results/RUN_MULTIALLELIC/summary_files/cross_strat_gRNAs/non_excision_guides/results'
excision_path = '/wynton/home/capra/gramey02/dnd_project/results/RUN_MULTIALLELIC/summary_files/cross_strat_gRNAs/excision_guides/results'
```

```python
# concatenate all files together
frames = []
file_list=[os.listdir(non_excision_path),os.listdir(excision_path)]
for files in tqdm(file_list): 
    for f in tqdm(files):
        stem = os.path.basename(f).removesuffix("_gRNAs.csv")   # 'SMARCA5_base_edits'
        gene, strategy = stem.split("_", 1)                      # 'SMARCA5', 'base_edits'

        if strategy=='excision':
            df = pd.read_csv(os.path.join(excision_path, f))
        else:
            df = pd.read_csv(os.path.join(non_excision_path, f))
        df["gene"] = gene
        df["strategy"] = strategy
        frames.append(df)

combined = pd.concat(frames, ignore_index=True)
combined["strategy"] = combined["strategy"].astype("category")
combined["gene"] = combined["gene"].astype("category")
```
